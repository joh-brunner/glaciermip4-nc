Load template csv file
This is the original file (downloaded as csv) from the Standardization-Document:
[Standardization Table](https://docs.google.com/spreadsheets/d/1L4y8xKgZjDjjA-d7mih33UkMP_v-xpoFGaShOb9iLnA/edit?gid=1539543782#gid=1539543782)


```python
import pandas as pd 
csv_template = "GlacierMIP4_output_comparison - standardized_outputs.csv"
csv_df = pd.read_csv(csv_template)

# Fix the data
# Drop empty rows
```

```python

idx = csv_df.index[csv_df.iloc[:, 0] == "Frontal ablation"][0]
# Add sign to frontal ablation
insert_at = idx + 4

# empty row with same columns
empty_row = pd.DataFrame([[1] * len(csv_df.columns)], columns=csv_df.columns)

# insert row
csv_df = pd.concat(
    [csv_df.iloc[:insert_at], empty_row, csv_df.iloc[insert_at:]],
    ignore_index=True
)


idx = csv_df.index[csv_df.iloc[:, 0] == "Refreezing"][0]
# Add sign to frontal ablation
insert_at = idx + 4

# empty row with same columns
empty_row = pd.DataFrame([[1] * len(csv_df.columns)], columns=csv_df.columns)

# insert row
csv_df = pd.concat(
    [csv_df.iloc[:insert_at], empty_row, csv_df.iloc[insert_at:]],
    ignore_index=True
)



csv_df = csv_df.dropna(subset=[csv_df.columns[1]])

glacier_mip_col = csv_df["GlacierMIP4_standardized"]
```

Reshape the GlacierMIP4_standardized column

```python
import numpy as np


s = glacier_mip_col

 # target number of columns
n_cols = 7

# Determine required total length
total = int(np.ceil(len(s) / n_cols) * n_cols)

# Repeat and cut to required length
values = np.resize(s.values, total)

# Reshape
arr2d = values.reshape(-1, n_cols)
df = pd.DataFrame(arr2d)

feature_names = ["var_short_name", "var_long_name", "unit", "sign", "resolution", "type", "comment"]
df.columns = feature_names

df.to_csv("out.csv")
```
