A quick cell to check whether a NetCDF file complies with the Climate and Forecasts (CF) Metadata Convention.

```python
import subprocess

nc_file = "OGGM_rgi01_MRI-ESM2-0_119_annual.nc"

result = subprocess.run(
    ["cfchecks", nc_file],
    capture_output=True,
    text=True,
)

print(result.stdout)
print(result.stderr)
```
